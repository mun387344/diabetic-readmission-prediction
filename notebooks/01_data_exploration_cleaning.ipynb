```python
# Missing Values


def count_missing_custom(col):
    return col.astype(str).isin(['?', 'None', 'nan', '']).sum()

missing_counts = df.apply(count_missing_custom)
missing_pct = (missing_counts / len(df)) * 100
missing_df = pd.DataFrame({'Count': missing_counts, 'Percentage': missing_pct})
print("\nColumns with Missing Values (sorted by %):")
print(missing_df[missing_df['Count'] > 0].sort_values(by='Percentage', ascending=False))

# 4. Duplicates
print(f"\nNumber of exact duplicate rows: {df.duplicated().sum()}")

# 5. Class Distribution (Target Variable)
print("\nClass Distribution ('readmitted'):")
print(df['readmitted'].value_counts())

# 6. Inspect Unusual Values
print("\nUnique values in 'age' (binned):", df['age'].unique())
print("Unique values in 'weight' (binned sample):", df['weight'].unique()[:5])
print("Unique values in 'gender':", df['gender'].unique())
print("Unique values in 'race':", df['race'].unique())
print("\nSummary Statistics:")
display(df.describe(include="all"))
```

### **Data Cleaning [Mohammed]**

```python

print(f"--- Initial Dataset Shape: {df.shape} ---\n")

# LOG ENTRY 1: Standardize Missing Values

missing_q_count = (df == '?').sum()
df.replace('?', np.nan, inplace=True)
print(f"[Log 1] Replaced {missing_q_count} '?' instances with np.nan.")

# LOG ENTRY 2: Drop Identifiers

ids_to_drop = ['encounter_id', 'patient_nbr']
df.drop(columns=ids_to_drop, inplace=True, errors='ignore')
print(f"[Log 2] Dropped {len(ids_to_drop)} identifier columns.")

# LOG ENTRY 3: Drop Extreme Missingness Columns

high_missing_cols = ['weight', 'max_glu_serum', 'A1Cresult']
df.drop(columns=high_missing_cols, inplace=True)

print(f"[Log 3] Dropped columns with >80% missingness.")

# LOG ENTRY 4: Impute Categorical Missingness

cat_cols = df.select_dtypes(include=['object']).columns
cat_missing_before = df[cat_cols].isna().sum()

# Fill missing categorical data with 'Unknown'
df[cat_cols] = df[cat_cols].fillna('Unknown')

print(f"[Log 4] Imputed categorical missing values with 'Unknown'. Affected observations:")
for col, count in cat_missing_before.items():
    if count > 0:
        print(f"   - {col}: {count} observations")

# LOG ENTRY 5: Encode Target Variable

if df['readmitted'].dtype == 'object':
    target_mapping = {'NO': 0, '>30': 1, '<30': 2}
    df['readmitted'] = df['readmitted'].map(target_mapping)
print(f"\n[Log 5] Encoded target variable 'readmitted' to numerical classes (0, 1, 2).")

# LOG ENTRY 6: Precautionary Numerical Imputation

num_cols = [col for col in df.columns if df[col].dtype != 'object' and col != 'readmitted']
num_missing_count = df[num_cols].isna().sum()

if num_missing_count.sum() > 0:
    for col in num_cols:
        df[col].fillna(df[col].median(), inplace=True)
    print(f"[Log 6] Imputed {num_missing_count.sum()} missing numerical values with median.")
else:
    print(f"[Log 6] Checked numerical columns: 0 missing values found.")

print(f"\n--- Final Cleaned Dataset Shape: {df.shape} ---")
if 'readmitted' in df.columns:
    print(f"Target Variable Distribution:\n{df['readmitted'].value_counts().sort_index()}")
```
