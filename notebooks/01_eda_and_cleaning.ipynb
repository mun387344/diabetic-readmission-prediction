### **Data Cleaning**

```python
# Audit columns that contain '?' vs native NaNs
q_marks = (df_raw == "?").sum()
native_nans = df_raw.isnull().sum()

missing_audit = pd.DataFrame({
    'question_mark_count': q_marks,
    'question_mark_pct': (q_marks / len(df_raw)) * 100,
    'native_nan_count': native_nans,
    'native_nan_pct': (native_nans / len(df_raw)) * 100
})

# Display all columns where either marker is present
missing_detected = missing_audit[(missing_audit['question_mark_count'] > 0) | (missing_audit['native_nan_count'] > 0)]
print(missing_detected.sort_values(by='question_mark_count', ascending=False))
```

Handling High Missingness & Clinical Category Preservation

```python
# Create our intentional working dataframe
df_processed = df_raw.copy()

# 1. Standardize '?' markers to true NaN across the entire dataframe
df_processed.replace("?", np.nan, inplace=True)

# 2. Treat absence of lab tests as an explicit clinical category ('None')
df_processed["A1Cresult"] = df_processed["A1Cresult"].fillna("None")
df_processed["max_glu_serum"] = df_processed["max_glu_serum"].fillna("None")

# 3. Impute high-missing administrative and clinical categories with an explicit token
df_processed["medical_specialty"] = df_processed["medical_specialty"].fillna("Unknown_Specialty")
df_processed["payer_code"] = df_processed["payer_code"].fillna("Missing_Payer")
df_processed["race"] = df_processed["race"].fillna("Unknown_Race")

# 4. Remove 'weight' deliberately due to >96% missingness
df_processed.drop(columns=["weight"], inplace=True)

print("Remaining columns count:", len(df_processed.columns))
print("\nValidation - A1Cresult categories:")
print(df_processed["A1Cresult"].value_counts())
print("\nValidation - max_glu_serum categories:")
print(df_processed["max_glu_serum"].value_counts())
```

Target Variable Creation & Verification

```python
# Map target: 1 if readmitted in <30 days, 0 otherwise (>30 or NO)
df_processed["target_readmit_30d"] = df_processed["readmitted"].apply(
    lambda x: 1 if x == "<30" else 0
)

# Validate the mapping against original categories
print("Cross-tabulation check (Original vs Binary Target):")
print(pd.crosstab(df_processed["readmitted"], df_processed["target_readmit_30d"], margins=True))

print("\nPositive Class Proportion (%):")
print(df_processed["target_readmit_30d"].value_counts(normalize=True) * 100)
```

High-Risk Demographic & Discharge Cleanup

```python
# 1. Drop the 3 invalid gender records
df_processed = df_processed[df_processed["gender"] != "Unknown/Invalid"]

# 2. Exclude deceased and hospice discharge dispositions (cannot be readmitted)
# 11: Expired
# 13: Hospice / home
# 14: Hospice / medical facility
# 19: Expired at home (hospice)
# 20: Expired in medical facility (hospice)
# 21: Expired, place unknown (hospice)
terminal_dispositions = [11, 13, 14, 19, 20, 21]
initial_rows = len(df_processed)
df_processed = df_processed[~df_processed["discharge_disposition_id"].isin(terminal_dispositions)]

removed_rows = initial_rows - len(df_processed)
print(f"Removed {removed_rows} terminal/deceased patient records.")
print(f"Current cleaned dataset shape: {df_processed.shape}")
```

Diagnosis Code Grouping (Clinical Feature Engineering)

```python
# Function to map raw ICD-9 codes to clinical disease categories
def map_icd9_to_category(code):
    if pd.isnull(code) or code == "Unknown":
        return "Other"
    
    code_str = str(code).strip()
    
    # Handle codes starting with 'V' or 'E' (Supplementary classifications)
    if code_str.startswith(("V", "E")):
        return "Other"
    
    # Extract the base numeric code (e.g., '250.8' -> 250)
    try:
        val = float(code_str.split(".")[0])
    except ValueError:
        return "Other"
    
    # Standard ICD-9 diagnostic disease grouping intervals
    if (390 <= val <= 459) or val == 785:
        return "Circulatory"
    elif (460 <= val <= 519) or val == 786:
        return "Respiratory"
    elif (520 <= val <= 579) or val == 787:
        return "Digestive"
    elif val == 250:
        return "Diabetes"
    elif (800 <= val <= 999):
        return "Injury"
    elif (710 <= val <= 739):
        return "Musculoskeletal"
    elif (580 <= val <= 629) or val == 788:
        return "Genitourinary"
    elif (140 <= val <= 239):
        return "Neoplasms"
    else:
        return "Other"

# Apply mapping to all three diagnostic variables
for col in ["diag_1", "diag_2", "diag_3"]:
    df_processed[f"{col}_group"] = df_processed[col].apply(map_icd9_to_category)

# Drop raw diagnosis columns to prevent high-cardinality leakage
df_processed.drop(columns=["diag_1", "diag_2", "diag_3"], inplace=True)

print("Validation - diag_1_group distribution:")
print(df_processed["diag_1_group"].value_counts())
print("\nCurrent Shape:", df_processed.shape)
```

Patient-Level Stratified Train-Test Split (Leakage Prevention)

```python
from sklearn.model_selection import train_test_split

# 1. Identify each patient's maximum target value (1 if they ever readmitted <30d, else 0)
patient_target_map = df_processed.groupby("patient_nbr")["target_readmit_30d"].max()

# 2. Split unique patients stratified by this patient-level target
train_patients, test_patients = train_test_split(
    patient_target_map.index,
    test_size=0.20,
    random_state=42,
    stratify=patient_target_map.values
)

train_patients_set = set(train_patients)
test_patients_set = set(test_patients)

# 3. Partition full encounter dataframe based on patient assignments
train_df = df_processed[df_processed["patient_nbr"].isin(train_patients_set)].copy()
test_df = df_processed[df_processed["patient_nbr"].isin(test_patients_set)].copy()

# 4. Critical Leakage Check: Ensure intersection of patient sets is strictly 0
patient_leakage = len(train_patients_set.intersection(test_patients_set))
print(f"Patient overlap between Train and Test sets: {patient_leakage}")

# 5. Class balance check across sets
print(f"\nTrain Encounter Count: {len(train_df)} | Positive Class Rate: {train_df['target_readmit_30d'].mean():.4f}")
print(f"Test Encounter Count:  {len(test_df)}  | Positive Class Rate: {test_df['target_readmit_30d'].mean():.4f}")
```
